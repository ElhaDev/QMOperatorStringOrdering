```python
def latex_coeffecient_parser(s: str, full_term: str = "") -> Fraction:
    """Helper method for term passer, reads a suspected positive part of the coeffecient and turns into a Fraction object. Throws exception otherwise.
    
    Args:
        s: the suspected fraction substring of the full_term e.g. "\frac{1}{3}", "3", etc.
        full_term: the full_term, to identify the term if an exception is raised

    Returms:
        coeff: the coeffecient as a Fraction object
    """
    s = s.strip()
    #check if fraction
    try:
        coeff = None
        if s[0:6] == "\\frac{": #if so s = '\frac{*}{*}'
            numerator, denominator = None, None
            
            i_frcb = int(s.index('}'))
            numerator = int(s[6: i_frcb])
            if s[i_frcb + 1] != '{': raise(Exception(f"Error reading coeffecient of {full_term}"))
            denominator = int(s[i_frcb + 2: -1])
            coeff = Fraction(numerator, denominator)
        else: #if so s= '7, 2, 3, etc.'
            coeff = Fraction(int(s),1)
        
        return coeff
    except Exception:
        raise(Exception(f"Error reading coeffecient of {full_term}"))
```

```python
#testing coeffecient_parser: check
test_frac_list = ["\\frac{1}{9}", "\\frac{-1}{9}", "\\frac{1}{-9}", "\\frac{0}{46}", "7"]
test_frac_answers = [Fraction(1,9), Fraction(-1,9), Fraction(-1,9), Fraction(0,1), Fraction(7,1)]
for i in range(len(test_frac_answers)):
    assert(latex_coeffecient_parser(test_frac_list[i], test_frac_list[i]) == test_frac_answers[i])
```
